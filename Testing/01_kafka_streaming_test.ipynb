```python
import json
kafka_connection_json = dbutils.secrets.get(scope='finguard-scope',key = 'kafka_connection_details')
kafka_config = json.loads(kafka_connection_json)
BOOTSTRAP_SERVERS = kafka_config['bootstrap_servers']
API_KEY = kafka_config['api_key']
API_SECRET = kafka_config['api_secret']
TOPIC_NAME = kafka_config['topic']
```

```python
jaas_config = f"kafkashaded.org.apache.kafka.common.security.plain.PlainLoginModule required username='{API_KEY}' password='{API_SECRET}';"
```

```python
sample_batch = (spark.read.format("kafka")
                    .option("kafka.bootstrap.servers", BOOTSTRAP_SERVERS)
                    .option("subscribe", TOPIC_NAME)
                    .option("kafka.security.protocol", "SASL_SSL")
                    .option("kafka.sasl.mechanism", "PLAIN")
                    .option("kafka.sasl.jaas.config",jaas_config)
                    .option("startingOffsets","earliest")
                    .load()
)
```

```python
sample_batch.count()
```

```python
display(sample_batch)
```

```python
# casting data type to string
from pyspark.sql.functions import col
casting_data_batch = sample_batch.select(col("key").cast("string"), col("value").cast("string"), col("topic"), col("partition"), col("offset"), col("timestamp"), col("timestampType"))
display(casting_data_batch)
```

```python
casting_data_batch.printSchema()
```

```python
# load the raw to bronze table 
casting_data_batch.write.saveAsTable("finguard.bronze.transactions_batch_test")
```

Streaming Table 

```python
streaming_df = (spark.readStream.format("kafka")
                    .option("kafka.bootstrap.servers", BOOTSTRAP_SERVERS)
                    .option("subscribe", TOPIC_NAME)
                    .option("kafka.security.protocol", "SASL_SSL")
                    .option("kafka.sasl.mechanism", "PLAIN")
                    .option("kafka.sasl.jaas.config",jaas_config)
                    .option("startingOffsets","earliest")
                    .load()
)
```

```python
from pyspark.sql.functions import col
parsed_streaming_df = streaming_df.select(col("key").cast("string"), col("value").cast("string"), col("topic"), col("partition"), col("offset"), col("timestamp"), col("timestampType"))
```

```python
streaming_query = (parsed_streaming_df.writeStream.format("delta")\
                    .outputMode("append")\
                    .option("checkpointLocation", "/Volumes/finguard/source/transaction/checkpoint/")\
                    .trigger(availableNow=True)\
                    .toTable("finguard.bronze.transactions_stream_test"))

print(streaming_query.id)
```
